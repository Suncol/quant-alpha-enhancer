# return_y.pkl data quality checks

This notebook profiles a date-by-stock return/label matrix. It is designed to be portable: set `RETURN_Y_PATH` to the pickle path on the current machine, or use the default `~/Downloads/lgbm_test/return_y.pkl`.

```python
from pathlib import Path
import json
import os

from analysis.return_y_quality_profile import profile_return_matrix

data_path = Path(os.environ.get("RETURN_Y_PATH", Path.home() / "Downloads" / "lgbm_test" / "return_y.pkl"))
summary = profile_return_matrix(data_path)
print(json.dumps({
    "shape": summary["shape"],
    "dates": summary["dates"],
    "schema": summary["schema"],
    "completeness": {
        k: summary["completeness"][k]
        for k in [
            "missing_pct",
            "inf_cells",
            "all_nan_rows_count",
            "all_nan_rows",
            "all_nan_columns_count",
            "all_nan_columns",
            "one_value_columns_count",
            "columns_lt_252_values",
        ]
    },
    "numeric_validity": {
        "summary": summary["numeric_validity"]["summary"],
        "threshold_counts": summary["numeric_validity"]["threshold_counts"],
        "top_abs_extremes": summary["numeric_validity"]["top_abs_extremes"][:10],
    },
}, ensure_ascii=False, indent=2, default=str))
```

```python
import pandas as pd

df = pd.read_pickle(data_path)
display(df.iloc[:5, :10])
display(pd.DataFrame(summary["temporal_profile"]["annual"]))
display(pd.DataFrame(summary["completeness"]["lowest_coverage_rows"]))
display(pd.DataFrame(summary["completeness"]["lowest_coverage_columns"]))
```
